# 3.2、Opencv和Numpy的安装

```python
import cv2 as cv
import matplotlib.pyplot as plt

img = cv.imread('xiaoren.png')
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

plt.figure()
plt.subplot(1, 2, 1)
plt.imshow(img[:, :, ::-1])
plt.subplot(1, 2, 2)
plt.imshow(gray, cmap='gray')
plt.show()
```

```python
print("彩色图形状：", img.shape)
print("灰度图形状：", gray.shape)
print("彩色图大小：", img.size)
print("灰度图大小：", gray.size)
print("彩色图维度：", img.ndim)
print("灰度图维度：", gray.ndim)
print("彩色图像素BGR值：", img[0, 0])
print("灰度图像素灰度值：", gray[0, 0])
```

```python
print("彩色图flags：", img.flags)
print("灰度图flags：", gray.flags)
```

# 3.4、Python程序运行计时程序

```python
import time

start_time = time.perf_counter()
time.sleep(1)
total = 0
for i in range(1000000):
    total += i
end_time = time.perf_counter()
elapsed_seconds = end_time - start_time
elapsed_milliseconds = elapsed_seconds * 1000
print(f"耗时: {elapsed_milliseconds:.2f} ms")
```

# 4.4、大模型的基础微调

#### 4.2PyCharm工程访问大模型文件实现测试1

```python
from transformers import AutoModelForCausalLM, AutoTokenizer

# 模型文件采用在线的方式进行
# model_name = "Qwen/Qwen2.5-0.5B-Instruct"
model_name = r"D:\Projects\DeepBule\models\Qwen2.5-0.5B-Instruct"

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    dtype="auto",
    device_map="auto"
)
tokenizer = AutoTokenizer.from_pretrained(model_name)

instruction = "Produce a list of the top 5 NHL players in 2021."
input_text = ""

messages = [
    {"role": "system", "content": "You are Qwen, created by Alibaba Cloud. You are a helpful assistant."},
    {"role": "user", "content": f"Instruction: {instruction}\nInput: {input_text}"}
]

text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)
model_inputs = tokenizer([text], return_tensors="pt").to(model.device)
generated_ids = model.generate(
    **model_inputs,
    max_new_tokens=512
)
generated_ids = [
    output_ids[len(input_ids):] for input_ids, output_ids in zip(model_inputs.input_ids, generated_ids)
]
response = tokenizer.batch_decode(generated_ids, skip_special_tokens=True)[0]

print("Model output:", response)
```

#### 5.5、加大epoch数量（500）, sample=False 进行微调

```python
from transformers import AutoModelForCausalLM, AutoTokenizer

#model_name = "Qwen/Qwen2.5-0.5B-Instruct"
model_name = r"D:\Projects\DeepBule\models\Qwen2.5-0.5B-Instruct-after-500"

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    dtype="auto",
    device_map="auto"
)
tokenizer = AutoTokenizer.from_pretrained(model_name)

instruction = "Produce a list of the top 5 NHL players in 2021."
input_text = ""

messages = [
    {"role": "system", "content": "You are Qwen, created by Alibaba Cloud. You are a helpful assistant."},
    {"role": "user", "content": f"Instruction: {instruction}\nInput: {input_text}"}
]

text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)
model_inputs = tokenizer([text], return_tensors="pt").to(model.device)
generated_ids = model.generate(
    **model_inputs,
    max_new_tokens=512,
    do_sample=False
)
generated_ids = [
    output_ids[len(input_ids):] for input_ids, output_ids in zip(model_inputs.input_ids, generated_ids)
]
response = tokenizer.batch_decode(generated_ids, skip_special_tokens=True)[0]

print("Model output:", response)
```
