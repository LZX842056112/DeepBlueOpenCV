```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

img = cv.imread("./images/xiaoren.png")
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
gray = np.float32(gray)
dst = cv.cornerHarris(gray, 2, 3, 0.04)
img[dst > 0.01 * dst.max()] = [255, 0, 255]
plt.imshow(img[:, :, ::-1])
plt.show()
```

# Shi-Tomasi角点检测

![shi](picture/shi.png)

![shi2](picture/shi2.png)

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

img = cv.imread("./images/xiaoren.png")
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
corners = cv.goodFeaturesToTrack(gray, 1000, 0.01, 10)

# 转换为整数坐标
corners = np.int0(corners)

for i in corners:
    x, y = i.ravel()
    cv.circle(img, (x, y), 2, (255, 0, 255), -1)

plt.imshow(img[:, :, ::-1])
plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

img = cv.imread("./images/xiaoren.png")
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

sift = cv.xfeatures2d.SIFT_create()
kp, des = sift.detectAndCompute(gray, None)
img = cv.drawKeypoints(img, kp, img, flags=cv.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)

plt.figure(figsize=(8, 6), dpi=100)
plt.imshow(img[:, :, ::-1])
plt.xticks([]), plt.yticks([])
plt.show()
```

```python
img = cv.imread("./images/xiaoren.png")
fast = cv.FastFeatureDetector_create(threshold=30)
kp = fast.detect(gray, None)
img2 = cv.drawKeypoints(img, kp, None,color=(0, 0, 255))

fast.setNonmaxSuppression(False)
kp = fast.detect(img, None)
img3 = cv.drawKeypoints(img, kp, None,color=(0, 0, 255))

plt.figure(figsize=(8, 6), dpi=100)
plt.subplot(1, 2, 1)
plt.imshow(img2[:, :, ::-1])
plt.subplot(1, 2, 2)
plt.imshow(img3[:, :, ::-1])
plt.show()
```

# 图像特征提取
scikit-image 是一个开源图像处理库。

```python
# pip install scikit-image
```

```python
import numpy as np
# import skimage as ski
from skimage import color as ski_color
from skimage import io as ski_io
from skimage import transform as ski_transform
from skimage import feature as ski_feature
```

```python
img = ski_io.imread("./images/xiaoren.png")  # 加载图像
img = ski_color.rgb2gray(img)  # 把 RGB 图转成灰度图
print("原图形状：", img.shape)

img = ski_transform.resize(img, (100, 300))
print("缩小图像：", img.shape)
```

（以下内容很少用，只需了解）

```python
# NOTE: 根据特征，可以建分类模型，只用于机器学习模型，不会用于深度学习模型

# 提取 HOG 特征
hog_feature = ski_feature.hog(img)
print(hog_feature.shape)
```

```python
# 提取 LBP 特征
img = np.uint8(img)

lbp_feature = ski_feature.local_binary_pattern(img, P=8, R=0.5).reshape(-1)
print(lbp_feature.shape)
```

```python
# 提取 Haar 特征
corner_harris_feature = ski_feature.corner_harris(img).reshape(-1)
print(corner_harris_feature.shape)
```
