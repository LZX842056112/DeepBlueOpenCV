```python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt
```

```python
img1 = cv.imread('../images/xiaoren.png')

kernel = np.ones((3, 3), np.float32) / 9
print(kernel)

img2 = cv.filter2D(img1, -1, kernel)

h, w, _ = img1.shape
w = int(w * 0.5)
h = int(h * 0.5)
img3 = cv.resize(img1, (w, h))
img4 = cv.resize(img2, (w, h))

w = int(w * 0.2)
h = int(h * 0.2)
img5 = cv.resize(img3, (w, h))
img6 = cv.resize(cv.filter2D(img4, -1, kernel), (w, h))

fig, axex = plt.subplots(2, 3, figsize=(10, 8), dpi=100)
axex[0, 0].imshow(cv.cvtColor(img1, cv.COLOR_BGR2RGB))
axex[0, 1].imshow(cv.cvtColor(img2, cv.COLOR_BGR2RGB))
axex[0, 2].imshow(cv.cvtColor(img3, cv.COLOR_BGR2RGB))
axex[1, 0].imshow(cv.cvtColor(img4, cv.COLOR_BGR2RGB))
axex[1, 1].imshow(img5[:, :, ::-1])
axex[1, 2].imshow(img6[:, :, ::-1])
plt.show()
```

```python
img1 = cv.imread('../images/xiaoren.png')

img2 = cv.GaussianBlur(img1, ksize=(5, 5), sigmaX=2.0, sigmaY=2.0)

plt.figure(figsize=(20, 10))

plt.subplot(2, 6, 1)
plt.imshow(cv.cvtColor(img1, cv.COLOR_BGR2RGB))  # 原图
plt.subplot(2, 6, 7)
plt.imshow(cv.cvtColor(img2, cv.COLOR_BGR2RGB))  # 高斯模糊

for i in range(5):
    h, w, _ = img1.shape  # 高度、宽度
    w = int(w * 0.5)
    h = int(h * 0.5)
    g_img = cv.GaussianBlur(img1, ksize=(5, 5), sigmaX=2.0, sigmaY=2.0)  # 高斯模糊
    img1 = cv.resize(img1, (w, h))  # 缩小图像
    img2 = cv.resize(g_img, (w, h))  # 高斯模糊 + 缩小图像
    plt.subplot(2, 6, i + 2)
    plt.imshow(cv.cvtColor(img1, cv.COLOR_BGR2RGB))
    plt.subplot(2, 6, i + 8)
    plt.imshow(cv.cvtColor(img2, cv.COLOR_BGR2RGB))

plt.show()
```

```python
img = cv.imread('../images/xiaoren.png')

noisy_img = np.random.normal(10, 10, (img.shape[0], img.shape[1], img.shape[2]))
noisy_img = np.clip(noisy_img, 0, 255).astype(np.uint8)

img = img + noisy_img
img = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

dst = cv.medianBlur(img, 5)

fig, axex = plt.subplots(1, 2, figsize=(10, 8), dpi=100)
axex[0].imshow(img, cmap='gray')
axex[1].imshow(dst, cmap='gray')
plt.show()
```

```python
img = cv.imread("../images/j.png", 0)

rows, cols = img.shape
for i in range(100):
    x = np.random.randint(cols)
    y = np.random.randint(rows)
    img[y, x] = 255

kernel = np.ones((5, 5), np.uint8)
dst = cv.morphologyEx(img, cv.MORPH_OPEN, kernel, 1)

dst1 = cv.erode(img, kernel, iterations=1)

dst2 = cv.dilate(img, kernel, iterations=1)

fig, axex = plt.subplots(2, 2, figsize=(10, 8), dpi=100)
axex[0, 0].imshow(img, cmap='gray')
axex[0, 1].imshow(dst, cmap='gray')
axex[1, 0].imshow(dst1, cmap='gray')
axex[1, 1].imshow(dst2, cmap='gray')
plt.show()
```

```python
img = cv.imread("../images/j.png", 0)
rows, cols = img.shape
# 加白色点
for i in range(100):
    x = np.random.randint(cols)
    y = np.random.randint(rows)
    img[y, x] = 255

# 加黑色点
for i in range(1000):
    x = np.random.randint(cols)
    y = np.random.randint(rows)
    img[y, x] = 0

kernel = np.ones((5, 5), np.uint8)
dst = cv.morphologyEx(img, cv.MORPH_CLOSE, kernel, 1)

dst1 = cv.dilate(img, kernel, iterations=1)

dst2 = cv.erode(img, kernel, iterations=1)

fig, axex = plt.subplots(2, 2, figsize=(10, 8), dpi=100)
axex[0, 0].imshow(img, cmap='gray')
axex[0, 1].imshow(dst, cmap='gray')
axex[1, 0].imshow(dst1, cmap='gray')
axex[1, 1].imshow(dst2, cmap='gray')
plt.show()
```

```python
img0 = cv.imread("../images/car.jpg")
img0 = cv.cvtColor(img0, cv.COLOR_BGR2GRAY)
img = cv.GaussianBlur(img0, (5, 5), 0)

kernel = np.ones((5, 5), np.uint8)

img1 = cv.morphologyEx(img, cv.MORPH_GRADIENT, kernel, 1)

ret, img2 = cv.threshold(img1, 127, 255, cv.THRESH_BINARY)

kernel = np.asarray([
    [1, 1, 1, 1, 1, 1, 1, 1]
])
img3 = cv.dilate(img2, kernel, iterations=4)

fig, axex = plt.subplots(2, 3, figsize=(10, 8), dpi=100)
axex[0, 0].imshow(img0, cmap='gray')
axex[0, 1].imshow(img, cmap='gray')
axex[0, 2].imshow(img1, cmap='gray')
axex[1, 0].imshow(img2, cmap='gray')
axex[1, 1].imshow(img3, cmap='gray')
plt.show()
```
