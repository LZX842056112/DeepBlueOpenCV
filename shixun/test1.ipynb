```python
import cv2 as cv

img = cv.imread('xiaoren.png', cv.IMREAD_GRAYSCALE)
ret, binary = cv.threshold(img, 127, 255, cv.THRESH_BINARY)
# 统计连通域信息
num_labels, labels, stats, centroids = cv.connectedComponentsWithStats(binary, connectivity=8)
print(f"找到 {num_labels - 1} 个连通域（不包括背景）")
# 遍历每个连通域（从1开始，0是背景）
for i in range(1, num_labels):
    # 获取连通域的统计数据
    x = stats[i, cv.CC_STAT_LEFT]  # 边界框左上角x坐标
    y = stats[i, cv.CC_STAT_TOP]  # 边界框左上角y坐标
    w = stats[i, cv.CC_STAT_WIDTH]  # 边界框宽度
    h = stats[i, cv.CC_STAT_HEIGHT]  # 边界框高度
    area = stats[i, cv.CC_STAT_AREA]  # 连通域面积
    cx, cy = centroids[i]  # 连通域质心坐标

    print(f"连通域 {i}:")
    print(f"  位置: ({x}, {y})")
    print(f"  尺寸: {w} x {h}")
    print(f"  面积: {area}")
    print(f"  质心: ({cx:.2f}, {cy:.2f})")
```

```python
import cv2 as cv
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['Microsoft YaHei', 'SimHei', 'SimSun']
plt.rcParams['axes.unicode_minus'] = False

img = cv.imread('xiaoren.png', cv.IMREAD_GRAYSCALE)
template = cv.imread('template.png', cv.IMREAD_GRAYSCALE)
h, w = template.shape
# 6种匹配方式
methods = [
    'cv.TM_CCOEFF',
    'cv.TM_CCOEFF_NORMED',
    'cv.TM_CCORR',
    'cv.TM_CCORR_NORMED',
    'cv.TM_SQDIFF',
    'cv.TM_SQDIFF_NORMED'
]
print("序号\t匹配方式\t\t\t左上角坐标\t\t\t相似度")
print("-" * 80)
# 创建图形用于显示所有结果
fig, axes = plt.subplots(3, 4, figsize=(16, 12))
# 遍历所有匹配方法
for idx, meth in enumerate(methods):
    img_copy = img.copy()
    # 获取匹配方法
    method = eval(meth)
    # 使用给定的方式进行模板匹配
    result = cv.matchTemplate(img_copy, template, method)
    # 从数据中查找全局最小值、最大值，以及对应的位置
    min_val, max_val, min_loc, max_loc = cv.minMaxLoc(result)
    # 根据匹配方法确定最佳匹配点
    if method in [cv.TM_SQDIFF, cv.TM_SQDIFF_NORMED]:
        top_left = min_loc
        similarity = min_val
    else:
        top_left = max_loc
        similarity = max_val
    # 基于左上角的坐标计算右下角的坐标
    bottom_right = (top_left[0] + w, top_left[1] + h)
    # 打印：序号、匹配方式、左上角的点和右下角的点、相似度
    print(f"{idx + 1}\t{meth:<20}\t{top_left}\t\t{similarity:.4f}")
    # 在原图上绘制匹配区域
    display_img = cv.cvtColor(img_copy, cv.COLOR_GRAY2BGR)
    cv.rectangle(display_img, top_left, bottom_right, (0, 0, 255), 2)

    row = idx // 2
    col = (idx % 2) * 2
    axes[row, col].imshow(result, cmap='jet')
    axes[row, col].set_title(f'{meth}\n匹配度热力图')
    axes[row, col].axis('off')
    axes[row, col + 1].imshow(cv.cvtColor(display_img, cv.COLOR_BGR2RGB))
    axes[row, col + 1].set_title(f'{meth}\n检测位置')
    axes[row, col + 1].axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

img = np.zeros((300, 300, 3), dtype=np.uint8)
cv.line(img, (50, 50), (250, 50), (255, 255, 255), 2)  # 水平线
cv.line(img, (150, 50), (150, 250), (255, 255, 255), 2)  # 垂直线
cv.line(img, (50, 250), (250, 50), (255, 255, 255), 2)  # 对角线

gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
# 边缘检测
edges = cv.Canny(gray, 50, 150)
# 概率霍夫变换检测直线段
lines = cv.HoughLinesP(edges, 1, np.pi / 180, threshold=50, minLineLength=30, maxLineGap=10)
# 在原图上绘制检测到的直线段
result_img = img.copy()
if lines is not None:
    for line in lines:
        x1, y1, x2, y2 = line[0]
        # 绘制线段，颜色为绿色 (0, 255, 0)，线宽为 2
        cv.line(result_img, (x1, y1), (x2, y2), (0, 255, 0), 2)
print(f"检测到 {len(lines) if lines is not None else 0} 条线段")

# 显示结果
plt.figure(figsize=(15, 5))
plt.subplot(131)
plt.imshow(cv.cvtColor(img, cv.COLOR_BGR2RGB))
plt.title('原始图像')
plt.axis('off')
plt.subplot(132)
plt.imshow(edges, cmap='gray')
plt.title('边缘检测结果')
plt.axis('off')
plt.subplot(133)
plt.imshow(cv.cvtColor(result_img, cv.COLOR_BGR2RGB))
plt.title('概率霍夫变换检测直线段')
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

img = np.zeros((300, 300, 3), dtype=np.uint8)
cv.line(img, (50, 50), (250, 50), (255, 255, 255), 2)  # 水平线
cv.line(img, (150, 50), (150, 250), (255, 255, 255), 2)  # 垂直线
cv.line(img, (50, 250), (250, 50), (255, 255, 255), 2)  # 对角线

gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

# 边缘检测
edges = cv.Canny(gray, 50, 150)

# 标准霍夫变换检测直线
lines = cv.HoughLines(edges, 1, np.pi / 180, threshold=100)

# 在原图上绘制检测到的直线
result_img = img.copy()
if lines is not None:
    for line in lines:
        rho, theta = line[0]
        a = np.cos(theta)
        b = np.sin(theta)
        x0 = a * rho
        y0 = b * rho
        x1 = int(x0 + 1000 * (-b))
        y1 = int(y0 + 1000 * (a))
        x2 = int(x0 - 1000 * (-b))
        y2 = int(y0 - 1000 * (a))
        cv.line(result_img, (x1, y1), (x2, y2), (0, 0, 255), 2)

print(f"检测到 {len(lines) if lines is not None else 0} 条直线")

# 显示结果
plt.figure(figsize=(15, 5))

plt.subplot(131)
plt.imshow(cv.cvtColor(img, cv.COLOR_BGR2RGB))
plt.title('原始图像')
plt.subplot(132)
plt.imshow(edges, cmap='gray')
plt.title('边缘检测结果')
plt.subplot(133)
plt.imshow(cv.cvtColor(result_img, cv.COLOR_BGR2RGB))
plt.title('霍夫变换检测直线')
plt.tight_layout()
plt.show()
```

```python
import cv2 as cv
import numpy as np

template = cv.imread('shubiao.jpg', cv.IMREAD_GRAYSCALE)
capture = cv.VideoCapture(0)


def multi_scale_template_matching(frame, template, threshold=0.5):
    # 多尺度模板匹配函数
    gray = cv.cvtColor(frame, cv.COLOR_BGR2GRAY)
    gray_h, gray_w = gray.shape
    t_h, t_w = template.shape

    best_match_val = -1
    best_match_loc = None
    best_scale = 1.0

    # 定义多个尺度进行匹配
    scales = []
    # 确定合适的尺度范围
    max_scale = min(gray_w / t_w, gray_h / t_h) * 0.9
    min_scale = max(0.1, max_scale * 0.3)  # 最小缩小到原图的10%或max_scale的30%
    # 创建几个不同的尺度
    for scale in np.linspace(min_scale, max_scale, 5):
        scales.append(scale)
    # 如果max_scale > 1，可以尝试放大模板
    if max_scale > 1.2:
        for scale in np.linspace(1.1, min(2.0, max_scale), 3):
            scales.append(scale)

    for scale in scales:
        # 调整模板大小
        scaled_w = int(t_w * scale)
        scaled_h = int(t_h * scale)
        # 确保缩放后的模板尺寸不超过视频帧
        if scaled_w > gray_w or scaled_h > gray_h:
            continue
        scaled_template = cv.resize(template, (scaled_w, scaled_h), interpolation=cv.INTER_AREA)
        # 模板匹配
        res = cv.matchTemplate(gray, scaled_template, cv.TM_CCOEFF_NORMED)
        # 查找最佳匹配位置
        min_val, max_val, min_loc, max_loc = cv.minMaxLoc(res)
        # 更新最佳匹配
        if max_val > best_match_val:
            best_match_val = max_val
            best_match_loc = max_loc
            best_scale = scale
            best_size = (scaled_w, scaled_h)
    return best_match_val, best_match_loc, best_scale, best_size


# 阈值
threshold = 0.5
while True:
    success, frame = capture.read()
    if not success:
        break
    # 使用多尺度模板匹配
    best_val, best_loc, best_scale, best_size = multi_scale_template_matching(frame, template, threshold)

    # 在图像上显示信息
    cv.putText(frame, f'Threshold: {threshold:.2f}', (10, 30), cv.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 2)
    cv.putText(frame, f'Best match: {best_val:.3f}', (10, 60), cv.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 2)
    cv.putText(frame, f'Scale: {best_scale:.2f}', (10, 90), cv.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 2)

    # 如果匹配度足够高，绘制矩形
    if best_val >= threshold and best_loc is not None:
        top_left = best_loc
        bottom_right = (top_left[0] + best_size[0], top_left[1] + best_size[1])
        # 绘制矩形
        cv.rectangle(frame, top_left, bottom_right, (0, 255, 0), 2)
        # 显示匹配分数
        cv.putText(frame, f'Score: {best_val:.3f}', (top_left[0], top_left[1] - 10),
                   cv.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)

    # 显示结果
    cv.imshow('Template Matching', frame)

    key = cv.waitKey(1) & 0xFF
    if key == ord('q'):
        break
    elif key == ord('+'):
        threshold = min(threshold + 0.05, 1.0)
    elif key == ord('-'):
        threshold = max(threshold - 0.05, 0.1)

capture.release()
cv.destroyAllWindows()
```
